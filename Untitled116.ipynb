```python
import os
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# ==================================================
# CONFIG
# ==================================================
BASE_DIR = "/content/drive/MyDrive/Luamata South"
FILE_NAME = "LUAMATA SOUTH SOILS DATA.xlsx"
SHEET_NAME = "Sheet1"

# ==================================================
# DATA LOADER FUNCTION (REUSABLE PIPELINE)
# ==================================================
def load_luamata_data(base_dir, file_name, sheet_name="Sheet1"):

    file_path = os.path.join(base_dir, file_name)

    # ----------------------------
    # Validate paths
    # ----------------------------
    if not os.path.exists(base_dir):
        raise FileNotFoundError(f"Folder not found: {base_dir}")

    print("📂 Files in directory:")
    print(os.listdir(base_dir))

    if not os.path.exists(file_path):
        raise FileNotFoundError(f"Excel file not found: {file_path}")

    # ----------------------------
    # Load data
    # ----------------------------
    df = pd.read_excel(file_path, sheet_name=sheet_name)

    # ----------------------------
    # Basic cleaning
    # ----------------------------
    df = df.dropna(subset=["Sample ID", "East", "North"]).copy()

    df = df[
        ~df["Comment"].astype(str).str.contains(
            r"STANDARD|BLANK", case=False, na=False
        )
    ].copy()

    # ----------------------------
    # Rename columns (standard schema)
    # ----------------------------
    df = df.rename(columns={
        "Sample ID": "sample_id",
        "East": "easting",
        "North": "northing",
        "Elevation": "elevation",
        "Depth": "depth",
        "Slope": "slope",
        "Colour": "colour",
        "Terrain": "terrain",
        "Comment": "comment"
    })

    # ----------------------------
    # Type cleaning
    # ----------------------------
    df["depth"] = pd.to_numeric(df["depth"], errors="coerce").fillna(30)

    print(f"✔ Primary samples loaded: {len(df)}")

    return df


# ==================================================
# ENCODER FUNCTION (SAFE + REUSABLE)
# ==================================================
def encode_categoricals(df, cols):
    encoders = {}

    for col in cols:
        le = LabelEncoder()
        df[f"{col}_code"] = le.fit_transform(df[col].fillna("UNK").astype(str))
        encoders[col] = le

    return df, encoders


# ==================================================
# RUN PIPELINE
# ==================================================
df = load_luamata_data(BASE_DIR, FILE_NAME)

df, encoders = encode_categoricals(df, ["colour", "terrain", "slope"])

# ==================================================
# FINAL CHECK
# ==================================================
print("\n✔ Dataset shape:", df.shape)
df.head()
```

```python
import os
import numpy as np
import pandas as pd

# ==================================================
# FEATURE EXPORT FUNCTION
# ==================================================
def export_features(
    df,
    feature_cols,
    out_dir="outputs",
    prefix="luamata_field",
    dropna=True,
    verbose=True
):
    """
    Export ML-ready feature matrix + cleaned dataset.
    """

    # --------------------------------------------------
    # 1. CREATE OUTPUT DIRECTORY
    # --------------------------------------------------
    os.makedirs(out_dir, exist_ok=True)

    # --------------------------------------------------
    # 2. CHECK COLUMNS
    # --------------------------------------------------
    missing = [c for c in feature_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing columns: {missing}")

    df_work = df.copy()

    # --------------------------------------------------
    # 3. CLEAN DATA
    # --------------------------------------------------
    for c in feature_cols:
        df_work[c] = pd.to_numeric(df_work[c], errors="coerce")

    if dropna:
        df_work = df_work.dropna(subset=feature_cols)

    # final safety cleanup
    df_work = df_work.dropna(subset=feature_cols)

    # --------------------------------------------------
    # 4. BUILD FEATURE MATRIX
    # --------------------------------------------------
    X = df_work[feature_cols].to_numpy(dtype=float)

    # --------------------------------------------------
    # 5. SAVE FILES
    # --------------------------------------------------
    npy_path = os.path.join(out_dir, f"{prefix}_features.npy")
    csv_path = os.path.join(out_dir, f"{prefix}_features.csv")
    xlsx_path = os.path.join(out_dir, f"{prefix}_clean.xlsx")

    np.save(npy_path, X)
    df_work.to_csv(csv_path, index=False)
    df_work.to_excel(xlsx_path, index=False)

    # --------------------------------------------------
    # 6. PRINT OUTPUT (THIS WAS YOUR ISSUE BEFORE)
    # --------------------------------------------------
    if verbose:
        print("\n==============================")
        print("✔ FEATURE EXPORT COMPLETE")
        print("==============================")
        print(f"Samples used : {len(df_work)}")
        print(f"Feature shape : {X.shape}")
        print(f"NPY file      : {npy_path}")
        print(f"CSV file      : {csv_path}")
        print(f"Excel file    : {xlsx_path}")
        print("==============================\n")

    # --------------------------------------------------
    # 7. RETURN RESULTS
    # --------------------------------------------------
    return {
        "X": X,
        "shape": X.shape,
        "npy": npy_path,
        "csv": csv_path,
        "excel": xlsx_path,
        "df_clean": df_work
    }


# ==================================================
# CALL FUNCTION (THIS IS REQUIRED!)
# ==================================================

feature_cols = [
    'easting',
    'northing',
    'elevation',
    'depth',
    'colour_code',
    'terrain_code',
    'slope_code'
]

result = export_features(df, feature_cols)

# quick confirmation
print("DONE ✔", result["shape"])
```

```python
def safe_correlation_plot(df, cols):
    import pandas as pd
    import seaborn as sns
    import matplotlib.pyplot as plt

    # Ensure numeric
    data = df[cols].apply(pd.to_numeric, errors='coerce')

    # Drop bad rows
    data = data.dropna()

    if data.empty:
        raise ValueError("No valid numeric data for correlation")

    corr = data.corr()

    plt.figure(figsize=(6,5))
    sns.heatmap(corr, annot=True, fmt=".2f")
    plt.title("Correlation Matrix")
    plt.show()

    return corr


# Use it
safe_correlation_plot(df, ['easting','northing','elevation','depth'])

```

```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import contextily as ctx
from shapely.geometry import Point

# ==================================================
# 1. CLEAN DATA FIRST (VERY IMPORTANT)
# ==================================================
def prepare_geo_data(df, x_col="easting", y_col="northing"):
    df = df.copy()

    df[x_col] = pd.to_numeric(df[x_col], errors="coerce")
    df[y_col] = pd.to_numeric(df[y_col], errors="coerce")

    df = df.dropna(subset=[x_col, y_col])

    return df


# ==================================================
# 2. SPATIAL PLOTTING FUNCTION
# ==================================================
def plot_spatial_maps(
    df,
    x_col="easting",
    y_col="northing",
    terrain_col="terrain",
    value_col="elevation",
    crs="EPSG:32735",
    figsize=(16, 8),
    cmap="terrain",
    save_path=None,
    show=True
):

    # --------------------------------------------------
    # 1. VALIDATE COLUMNS
    # --------------------------------------------------
    required = [x_col, y_col, terrain_col, value_col]
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise ValueError(f"Missing columns: {missing}")

    # --------------------------------------------------
    # 2. CLEAN DATA (CRITICAL FIX)
    # --------------------------------------------------
    df = prepare_geo_data(df, x_col, y_col)

    # --------------------------------------------------
    # 3. CREATE GEODATAFRAME
    # --------------------------------------------------
    gdf = gpd.GeoDataFrame(
        df.copy(),
        geometry=[Point(xy) for xy in zip(df[x_col], df[y_col])],
        crs=crs
    )

    # --------------------------------------------------
    # 4. CHECK IF DATA EXISTS
    # --------------------------------------------------
    if len(gdf) == 0:
        raise ValueError("No valid data to plot after cleaning")

    # --------------------------------------------------
    # 5. CREATE PLOTS
    # --------------------------------------------------
    fig, ax = plt.subplots(1, 2, figsize=figsize)

    # ---- TERRAIN MAP ----
    gdf.plot(
        ax=ax[0],
        column=terrain_col,
        categorical=True,
        legend=True,
        markersize=30,
        edgecolor="black"
    )
    ax[0].set_title("Terrain Map")

    # ---- ELEVATION MAP ----
    sc = ax[1].scatter(
        gdf.geometry.x,
        gdf.geometry.y,
        c=gdf[value_col],
        cmap=cmap,
        s=25
    )

    plt.colorbar(sc, ax=ax[1], label=value_col)
    ax[1].set_title("Elevation Map")

    # --------------------------------------------------
    # 6. BASEMAP (SAFE)
    # --------------------------------------------------
    for a in ax:
        try:
            ctx.add_basemap(
                a,
                crs=gdf.crs,
                source=ctx.providers.OpenStreetMap.Mapnik
            )
        except Exception:
            pass  # ignore if no internet

    plt.tight_layout()

    # --------------------------------------------------
    # 7. SAVE OUTPUT
    # --------------------------------------------------
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches="tight")

    # --------------------------------------------------
    # 8. SHOW PLOT (FORCE DISPLAY)
    # --------------------------------------------------
    if show:
        plt.show()
    else:
        plt.close()

    return gdf, fig


# ==================================================
# 3. RUN THE FUNCTION (THIS IS REQUIRED)
# ==================================================

# FIX DATA FIRST (important!)
df = prepare_geo_data(df)

gdf, fig = plot_spatial_maps(
    df,
    save_path="spatial_maps.png"
)

print("DONE ✔ Plot generated successfully")
```

```python
minx, maxx = df.easting.min(), df.easting.max()
miny, maxy = df.northing.min(), df.northing.max()
# Round to 200m
minx = np.floor(minx/200)*200
maxx = np.ceil(maxx/200)*200
miny = np.floor(miny/200)*200
maxy = np.ceil(maxy/200)*200

x_edges = np.arange(minx, maxx+200, 200)
y_edges = np.arange(miny, maxy+200, 200)
cells = []
for xc in x_edges[:-1]:
    for yc in y_edges[:-1]:
        cells.append({'easting': xc+100, 'northing': yc+100, 'cell_center_x': xc+100, 'cell_center_y': yc+100})
grid = pd.DataFrame(cells)

def cell_sampled(cx, cy, samples, radius=100):
    distances = np.sqrt((samples.easting - cx)**2 + (samples.northing - cy)**2)
    return np.any(distances <= radius)

grid['sampled'] = grid.apply(lambda r: cell_sampled(r['cell_center_x'], r['cell_center_y'], df), axis=1)
coverage = 100 * grid['sampled'].sum() / len(grid)
print(f"Grid coverage: {coverage:.1f}% ({grid['sampled'].sum()}/{len(grid)} cells)")

# Map coverage
plt.figure(figsize=(10,10))
plt.scatter(grid[grid.sampled]['cell_center_x'], grid[grid.sampled]['cell_center_y'], c='green', s=50, label='Sampled cell')
plt.scatter(grid[~grid.sampled]['cell_center_x'], grid[~grid.sampled]['cell_center_y'], c='red', s=50, label='Missed')
plt.scatter(df.easting, df.northing, c='blue', s=10, alpha=0.5, label='Sample')
plt.xlabel('Easting')
plt.ylabel('Northing')
plt.title(f'200m Grid Coverage – {coverage:.1f}%')
plt.legend()
plt.savefig('grid_coverage.png', dpi=200)
plt.show()
```

```python
feat = ['easting','northing','elevation']
Xc = df[feat].values
scaler = StandardScaler()
Xcs = scaler.fit_transform(Xc)

inertias = []
for k in range(1,8):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(Xcs)
    inertias.append(km.inertia_)
plt.plot(range(1,8), inertias, 'o-')
plt.xlabel('k')
plt.ylabel('Inertia')
plt.title('Elbow for K‑means')
plt.savefig('kmeans_elbow.png')
plt.show()

kmeans = KMeans(n_clusters=3, random_state=42)
df['cluster'] = kmeans.fit_predict(Xcs)

# Map clusters
gdf['cluster'] = df['cluster']
fig, ax = plt.subplots(figsize=(10,10))
gdf.plot(ax=ax, column='cluster', categorical=True, legend=True, markersize=30, cmap='viridis')
ctx.add_basemap(ax, crs=gdf.crs)
ax.set_title('K‑means Clusters (location + elevation)')
plt.savefig('cluster_map.png', dpi=200)
plt.show()
```

```python
pca_feat = ['easting','northing','elevation','depth','colour_code','terrain_code','slope_code']
Xp = df[pca_feat].values
Xps = StandardScaler().fit_transform(Xp)
pca = PCA(n_components=2)
pcs = pca.fit_transform(Xps)
df['PC1'] = pcs[:,0]
df['PC2'] = pcs[:,1]

print("Explained variance:", pca.explained_variance_ratio_)
loadings = pd.DataFrame(pca.components_.T, columns=['PC1','PC2'], index=pca_feat)
print("\nLoadings:\n", loadings)

# Plot PC1 vs PC2
plt.figure(figsize=(8,6))
sc = plt.scatter(df['PC1'], df['PC2'], c=df['terrain_code'], cmap='tab10')
plt.colorbar(sc, label='Terrain code')
plt.xlabel(f'PC1 ({100*pca.explained_variance_ratio_[0]:.1f}%)')
plt.ylabel(f'PC2 ({100*pca.explained_variance_ratio_[1]:.1f}%)')
plt.title('PCA – All Field Variables')
plt.savefig('pca_scores.png')
plt.show()
```

```python
iso = IsolationForest(contamination=0.05, random_state=42)
df['outlier'] = iso.fit_predict(Xps)  # -1 = outlier
outliers = df[df.outlier == -1]
print(f"Outliers: {len(outliers)}")

# Map outliers
gdf['outlier'] = df['outlier']
fig, ax = plt.subplots(figsize=(10,10))
gdf[gdf.outlier==1].plot(ax=ax, color='lightblue', markersize=20, label='Inlier')
gdf[gdf.outlier==-1].plot(ax=ax, color='red', markersize=80, edgecolor='black', label='Outlier')
ctx.add_basemap(ax, crs=gdf.crs)
ax.legend()
ax.set_title('Isolation Forest Outliers (field parameters)')
plt.savefig('outlier_map.png', dpi=200)
plt.show()
```
